# Identifikasi Dataset Klasifikasi Teks

**Tujuan:** mengidentifikasi fitur dan target, mengubah data ke DataFrame, menghitung jumlah data dan variabel, menentukan jenis data (teks atau image), menjelaskan sumber dataset, lalu menyimpulkan jenis klasifikasinya.

Ada 3 dataset, masing-masing mewakili satu jenis klasifikasi:

| No | Judul Dataset | Jenis Klasifikasi |
|---|---|---|
| 1 | Klasifikasi SMS Spam | Biner |
| 2 | Klasifikasi Topik Berita | Multikelas |
| 3 | Klasifikasi Aspek Ulasan Toko Online | Multilabel |

**Penjelasan istilah:**
- **Fitur (X)**: data masukan yang dipakai model untuk belajar (di sini berupa kalimat teks).
- **Target / Label (y)**: jawaban yang ingin diprediksi model.
- **DataFrame**: tabel pandas berisi baris (data) dan kolom (variabel).


```python
import pandas as pd
pd.set_option('display.max_colwidth', 90)
```

---
# Dataset 1: Klasifikasi SMS Spam

## 1.1 Keterangan Dataset (Sumber)
**Penjelasan:** bagian ini menjelaskan dari mana dataset diambil dan isinya tentang apa, supaya jelas asal-usul datanya.

- **Judul:** Klasifikasi SMS Spam
- **Sumber:** dibuat sendiri (data sintetis) menggunakan Python, disimpan sebagai file `1_biner_sms_spam.csv`
- **Isi:** kumpulan pesan SMS berbahasa Indonesia yang diberi label spam atau bukan spam.

## 1.2 Ubah ke DataFrame
**Penjelasan:** file CSV dibaca dengan `pd.read_csv()` sehingga menjadi DataFrame, yaitu tabel yang mudah dianalisis. Setelah itu kita lihat 5 baris pertama untuk memastikan datanya terbaca benar.

```python
df1 = pd.read_csv('1_biner_sms_spam.csv')
df1.head()
```

## 1.3 Jumlah Data dan Variabel
**Penjelasan:** `shape` menampilkan (jumlah baris, jumlah kolom). **Jumlah baris = jumlah data**, **jumlah kolom = jumlah variabel**. Kolom `id` hanya nomor urut, bukan fitur maupun label.

```python
print('Jumlah data    :', df1.shape[0])
print('Jumlah variabel:', df1.shape[1])
print('Nama variabel  :', list(df1.columns))
df1.info()
```

## 1.4 Identifikasi Fitur dan Target
**Penjelasan:** kolom dipisahkan menjadi **fitur (X)** yaitu masukan model, dan **target (y)** yaitu label yang diprediksi.

- Fitur: teks
- Target: `label` (spam atau bukan_spam)

```python
fitur = ['teks']
target = ['label']

X = df1[fitur]
y = df1[target]

print('Fitur (X):', fitur)
print('Target (y):', target)
display(y.head())
```

## 1.5 Distribusi Label
**Penjelasan:** menghitung banyaknya data pada setiap label. Ini dipakai untuk menentukan jumlah kelas dan melihat apakah datanya seimbang.

```python
print(y[target[0]].value_counts())
```

## 1.6 Jenis Data: Teks atau Image
**Penjelasan:** jenis data ditentukan dari isi kolom fitur. Kolom `teks` berisi kalimat, jadi dataset ini berupa **data teks**, bukan image.

```python
print('Tipe kolom fitur:', df1[fitur].dtypes.iloc[0])
print('Contoh isi fitur :', df1[fitur].iloc[0, 0])
print('Kesimpulan jenis data: TEKS')
```

## 1.7 Kesimpulan Jenis Klasifikasi
**Penjelasan:** jenis klasifikasi ditentukan dari label.
- **Biner**: hanya 2 kelas.
- **Multikelas**: lebih dari 2 kelas, tiap data hanya punya 1 label.
- **Multilabel**: tiap data boleh punya lebih dari 1 label sekaligus.

**Kesimpulan:** label hanya memiliki 2 kelas (spam dan bukan_spam), sehingga ini adalah **Klasifikasi Biner**.

```python
n_kelas = df1[target[0]].nunique()
print('Jumlah kelas:', n_kelas)
print('Jenis klasifikasi:', 'BINER' if n_kelas == 2 else 'MULTIKELAS')
```

# Dataset 3: Klasifikasi Aspek Ulasan Toko Online

## 3.1 Keterangan Dataset (Sumber)
**Penjelasan:** bagian ini menjelaskan dari mana dataset diambil dan isinya tentang apa, supaya jelas asal-usul datanya.

- **Judul:** Klasifikasi Aspek Ulasan Toko Online
- **Sumber:** dibuat sendiri (data sintetis) menggunakan Python, disimpan sebagai file `3_multilabel_ulasan.csv`
- **Isi:** kumpulan ulasan toko online berbahasa Indonesia. Setiap ulasan menandai aspek yang dibahas (harga, kualitas, pengiriman, pelayanan) dengan nilai 1 atau 0.

## 3.2 Ubah ke DataFrame
**Penjelasan:** file CSV dibaca dengan `pd.read_csv()` sehingga menjadi DataFrame, yaitu tabel yang mudah dianalisis. Setelah itu kita lihat 5 baris pertama untuk memastikan datanya terbaca benar.

```python
df3 = pd.read_csv('3_multilabel_ulasan.csv')
df3.head()
```

## 3.3 Jumlah Data dan Variabel
**Penjelasan:** `shape` menampilkan (jumlah baris, jumlah kolom). **Jumlah baris = jumlah data**, **jumlah kolom = jumlah variabel**. Kolom `id` hanya nomor urut, bukan fitur maupun label.

```python
print('Jumlah data    :', df3.shape[0])
print('Jumlah variabel:', df3.shape[1])
print('Nama variabel  :', list(df3.columns))
df3.info()
```

## 3.4 Identifikasi Fitur dan Target
**Penjelasan:** kolom dipisahkan menjadi **fitur (X)** yaitu masukan model, dan **target (y)** yaitu label yang diprediksi.

- Fitur: ulasan
- Target: 4 kolom (`harga`, `kualitas`, `pengiriman`, `pelayanan`) bernilai 0 atau 1

```python
fitur = ['ulasan']
target = ['harga', 'kualitas', 'pengiriman', 'pelayanan']

X = df3[fitur]
y = df3[target]

print('Fitur (X):', fitur)
print('Target (y):', target)
display(y.head())
```

## 3.5 Distribusi Label
**Penjelasan:** menghitung banyaknya data pada setiap label. Ini dipakai untuk menentukan jumlah kelas dan melihat apakah datanya seimbang.

```python
print(y.sum())
```

## 3.6 Jenis Data: Teks atau Image
**Penjelasan:** jenis data ditentukan dari isi kolom fitur. Kolom `ulasan` berisi kalimat, jadi dataset ini berupa **data teks**, bukan image.

```python
print('Tipe kolom fitur:', df3[fitur].dtypes.iloc[0])
print('Contoh isi fitur :', df3[fitur].iloc[0, 0])
print('Kesimpulan jenis data: TEKS')
```

## 3.7 Kesimpulan Jenis Klasifikasi
**Penjelasan:** jenis klasifikasi ditentukan dari label.
- **Biner**: hanya 2 kelas.
- **Multikelas**: lebih dari 2 kelas, tiap data hanya punya 1 label.
- **Multilabel**: tiap data boleh punya lebih dari 1 label sekaligus.

**Kesimpulan:** satu ulasan bisa memiliki beberapa aspek sekaligus (lebih dari satu angka 1), sehingga ini adalah **Klasifikasi Multilabel**.

```python
jumlah_label = df3[target].sum(axis=1)
print('Jumlah kolom label:', len(target))
print('Data dengan lebih dari 1 label:', (jumlah_label > 1).sum())
print('Jenis klasifikasi:', 'MULTILABEL' if (jumlah_label > 1).any() else 'MULTIKELAS / BINER')
```
