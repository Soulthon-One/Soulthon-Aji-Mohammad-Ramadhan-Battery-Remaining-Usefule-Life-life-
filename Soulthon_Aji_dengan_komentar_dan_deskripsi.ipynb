## Bagian 12 — Menggabungkan observations dan metadata

**Deskripsi:** `.merge()` menggabungkan dua tabel berdasarkan kolom `battery_id`. `how='left'` mempertahankan seluruh baris observations, lalu menambahkan metadata yang cocok. Hasil gabungan disimpan dalam `df`; `.shape` dan `.head()` dipakai untuk memeriksa hasilnya.

## Bagian 19 — Membuat heatmap korelasi

**Deskripsi:** `select_dtypes(include=np.number)` memilih kolom numerik. `.corr()` menghitung korelasi antar kolom numerik, kemudian `sns.heatmap()` menampilkannya sebagai peta warna. Nilai mendekati 1 menunjukkan hubungan linear positif, mendekati -1 negatif, dan mendekati 0 hubungan linear yang lemah.

**Cara membaca hasil:** Perhatikan output yang muncul setelah sel dijalankan. Nilai aktualnya bergantung pada isi file CSV yang digunakan.

```python
# Hitung korelasi kolom numerik dan visualisasikan dengan heatmap.
numeric_df = df.select_dtypes(
    include=np.number
)

correlation = numeric_df.corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix")

plt.show()
```

## Bagian 22 — Mengubah kategori menjadi kolom numerik (one-hot encoding)

**Deskripsi:** `pd.get_dummies()` mengubah kategori pada chemistry, manufacturer, climate_zone, dan usage_strategy menjadi kolom indikator 0/1 agar dapat digunakan model machine learning. `drop_first=True` menghilangkan satu kategori acuan pada tiap fitur untuk mengurangi redundansi. Ukuran dan contoh hasil ditampilkan.

**Cara membaca hasil:** Perhatikan output yang muncul setelah sel dijalankan. Nilai aktualnya bergantung pada isi file CSV yang digunakan.

```python
# Lakukan one-hot encoding untuk fitur kategorikal yang dipilih.
df_processed = pd.get_dummies(
    df,
    columns=[
        "chemistry",
        "manufacturer",
        "climate_zone",
        "usage_strategy"
    ],
    drop_first=True
)

print("Ukuran data setelah encoding:")
print(df_processed.shape)

display(df_processed.head())
```

## Bagian 23 — Menghitung total missing value pada data hasil encoding

**Deskripsi:** `.isnull().sum().sum()` menghitung seluruh nilai kosong di semua kolom `df_processed`. Hasil nol berarti tidak ditemukan missing value pada tabel tersebut; jika lebih dari nol, perlu ditangani sebelum pelatihan model.

**Cara membaca hasil:** Perhatikan output yang muncul setelah sel dijalankan. Nilai aktualnya bergantung pada isi file CSV yang digunakan.

```python
# Hitung total missing value pada data yang sudah diproses.
print(
    "Total missing value:",
    df_processed.isnull().sum().sum()
)
```

## Bagian 24 — Mengambil ID untuk train, validation, dan test

**Deskripsi:** Kolom `battery_id` dari setiap file pembagian diubah menjadi list Python menggunakan `.tolist()`. Panjang setiap list ditampilkan untuk memeriksa jumlah ID di masing-masing kelompok.

**Cara membaca hasil:** Perhatikan output yang muncul setelah sel dijalankan. Nilai aktualnya bergantung pada isi file CSV yang digunakan.

```python
# Ubah daftar ID train/validation/test menjadi list.
train_id = train_batteries["battery_id"].tolist()

validation_id = validation_batteries[
    "battery_id"
].tolist()

test_id = test_batteries[
    "battery_id"
].tolist()

print("Jumlah battery train      :", len(train_id))
print("Jumlah battery validation :", len(validation_id))
print("Jumlah battery test       :", len(test_id))
```

## Bagian 25 — Membentuk dataset train, validation, dan test

**Deskripsi:** `.isin()` memilih baris `df_processed` yang ID baterainya ada di daftar masing-masing kelompok. `.copy()` membuat salinan DataFrame hasil filter. Ukuran tiap kelompok ditampilkan untuk memeriksa hasil pemisahan.

**Cara membaca hasil:** Perhatikan output yang muncul setelah sel dijalankan. Nilai aktualnya bergantung pada isi file CSV yang digunakan.

```python
# Pisahkan data berdasarkan daftar ID tiap kelompok.
train_data = df_processed[
    df_processed["battery_id"].isin(train_id)
].copy()

validation_data = df_processed[
    df_processed["battery_id"].isin(validation_id)
].copy()

test_data = df_processed[
    df_processed["battery_id"].isin(test_id)
].copy()

print("Train      :", train_data.shape)
print("Validation :", validation_data.shape)
print("Test       :", test_data.shape)
```

## Bagian 26 — Memeriksa variabel pembagian data yang tersedia

**Deskripsi:** `globals()` berisi nama variabel yang tersedia pada sesi Python. List comprehension menyaring nama yang mengandung kata train, validation, atau test. Ini hanya memeriksa nama variabel di memori, bukan memvalidasi isi atau kualitas dataset.

**Cara membaca hasil:** Perhatikan output yang muncul setelah sel dijalankan. Nilai aktualnya bergantung pada isi file CSV yang digunakan.

```python
# Tampilkan nama variabel yang mengandung train, validation, atau test.
print("Variabel yang tersedia:")
print([name for name in globals() if "train" in name.lower()
      or "validation" in name.lower()
      or "test" in name.lower()])
```
